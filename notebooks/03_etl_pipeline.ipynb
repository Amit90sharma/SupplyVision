# What is ETL?

ETL stands for:

**E**xtract

**T**ransform

**L**oad

It is the process of collecting raw data, cleaning and transforming it into a structured format, and loading it into a system where it can be analyzed.

Every business intelligence solution relies on ETL because dashboards are only as reliable as the data behind them.

# ETL in SupplyVision

Our project follows a batch ETL workflow.

```
Raw CSV Files
        │
        ▼
Dataset Audit
        │
        ▼
Data Quality Validation
        │
        ▼
Data Transformation
        │
        ▼
Feature Engineering
        │
        ▼
Date Dimension
        │
        ▼
Processed Data
        │
        ▼
MySQL Data Warehouse
        │
        ▼
SQL Analytics
        │
        ▼
Power BI Dashboard
```

Each stage improves the quality and usability of the data before it is consumed by business users.

# Batch ETL vs Real-Time ETL

| Batch ETL | Real-Time ETL |
|------------|---------------|
| Runs on schedule | Runs continuously |
| Easier to maintain | More complex |
| Suitable for historical analysis | Suitable for live dashboards |
| Lower infrastructure cost | Higher infrastructure cost |

SupplyVision uses **Batch ETL**, which is the most common approach for business reporting and analytics.

# ETL Components

## Extract

Read data from CSV files.

## Transform

- Validate data
- Standardize data types
- Create new features
- Generate date dimension
- Prepare analytical datasets

## Load

Store processed datasets for:

- MySQL
- SQL Analytics
- Power BI

# Why Not Load Raw Data?

Loading raw data directly into dashboards creates several problems:

- Inconsistent data types
- Duplicate logic across reports
- Slower dashboards
- Poor maintainability
- Difficult debugging

Instead, the ETL pipeline produces a clean and consistent analytical dataset that can be reused by multiple reports.

# SupplyVision ETL Architecture

# Business Value

The ETL pipeline ensures that:

- Business users work with trusted data.
- Reports remain consistent.
- KPIs are calculated from validated datasets.
- Future data loads follow the same process.
- Analytics can scale as more data becomes available.
